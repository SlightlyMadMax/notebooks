### Семинар 2.1 Основы Matplotlib
`matplotlib.pyplot` — стандарт для графиков.

Основные элементы: `plot`, `xlabel`, `ylabel`, `title`, `legend`, `grid`.

```python
# Базовый график

import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(-2, 2, 25)
z = np.sin(3*x)

#x1=np.linspace(-10,10,200)
#plt.plot(x1, np.sin(x1), 'r-')
#plt.axis([-5, 5, -2, 2])      # дапазон вывода по осям

plt.plot(x, z)
plt.show()
```

```python
# Подписи, титул, легенда, сетка...
# + несколько графиков на одном поле

x=np.linspace(-10,10,200)
plt.plot(x, np.sin(x), 'r:',label="one" )
plt.plot(x, np.sin(x+np.pi/2)/2, 'g',label="two")
plt.title('My Graph')
plt.xlabel('Х')
plt.ylabel('значение Y')
plt.legend()
plt.grid()         #  включить сетку
plt.show()
```

```python
help(plt.legend)
```

```python
# Логарифмический масштаб по осям.
x0 = np.linspace(0,2,100)
x = np.logspace(0,2,100)

z1 = np.exp(x/10)
z2 = np.sin(x)
y= z1+z2

plt.plot(x0,y,'.-r')
plt.title("lin-lin")
plt.show()

plt.loglog(x, y, '.-')
plt.title("log-log")
plt.show()

plt.semilogy(x, y, '.-')
plt.title("lin-log (semilogy)")
plt.show()
```

```python
# Subplot позволяет разбить графическое окно на несколько областей.
# формат: subplot(nplot,nrows,ncols)
# три параметра: nrows-количество строк, ncols - количество столбцов,
# nplot - номер области (от 1 до nrows*ncols).
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(-10, 10.01, 0.01)     # от -10 до 10.01 с шагом 0.01 (>1000 точек)
t = np.linspace(-10, 11, 20)        # от -10 до 11: 20 точек

#subplot 1
plt.subplot(221)
plt.plot(x, np.sin(x))
plt.title('sin(x)')
plt.grid()

#subplot 2
plt.subplot(222)
plt.plot(x, np.cos(x), 'g')
plt.axis('equal')           # одинаковый масштаб по осям Х и Y
plt.grid()
plt.title('cos(x)')

#subplot 3
plt.subplot(223)
plt.plot(x, x**2, t, t**2, 'ro')    # 2 графика, первый линией, второй красными кругами
plt.title('$x^2$')

#subplot 4
plt.subplot(224)
y=np.sin(t/3)
plt.errorbar(t, y, xerr=0.02, yerr=0.5)
plt.title('sin(t/3)')
plt.grid()

#plt.tight_layout()         # автоматически настраивает положение элементов (меток осей, заголовков),
                           # чтобы они не перекрывались и не вылезали за пределы окна фигуры.

# Или можно расположить области с заданными отступами
#plt.subplots_adjust(wspace=0.3, hspace=0.5)
#plt.subplots_adjust(left=0.1, right=0.95, bottom=0.1, top=0.95, wspace=0.2, hspace=0.4)

plt.show()
```

```python
# Полярная система координат (опц)
import numpy as np
import matplotlib.pyplot as plt

plt.subplot(111, polar=True)

phi = np.arange(0, 2*np.pi, 0.01)   # угол phi - массив от от 0 до 2*pi с шагом 0.01
rho = 2*phi                         # расстояние от центра 2*phi
plt.plot(phi, rho)
plt.show()
```

```python
# Параметрическая форма (опц)
# Окружность: x(t) = r*sin(t), y(t) = r*cos(t), где t от 0 до 2pi )

import numpy as np
import matplotlib.pyplot as plt

t = np.arange(0, 2*np.pi, 0.01)          # угол t от 0 до 2pi с шагом 0.01
r = 4                                    # радиус 4

plt.plot(r*np.sin(t), r*np.cos(t), lw=4) # x и у задаем как numpy функции от t
#plt.axis('equal')                        # масштаб осей Х и У одинаковый (чтобы круг не был овалом)

plt.show()
```

```python
# Метод создания графика, объединяющий инициализацию области для
# рисунка (fig) и осей (ax), на которых впоследствии будут отображаться данные.
# (Это тоже самое, что и базоый вариант, но бывает удобнее)

# Пример_1

x = np.linspace(-2, 2, 100)
z = np.sin(3*x)

fig, ax = plt.subplots()
#fig.set_size_inches(10, 3)     # Изменение размера холста

fig.figsize=(8, 6)            # Задает размер холста

ax.plot(x, np.sin(3*x),'.g')

ax.set_title('График')
ax.set_xlabel('Oсь Х')
ax.set_ylabel('Ось Y')

plt.show()
fig.savefig('plot.png')        # сохранить в файл (! в /content на Сolab)
```

```python
# Пример_2
# subplots, cетка 2×2:

x = np.linspace(-1, 3.0, 500)
fig, axes = plt.subplots(2, 2, figsize=(8, 6))

# axes — двумерный массив
axes[0, 0].plot(x, np.sin(x))
axes[0, 0].set_title('sin(x)')
axes[0, 0].set_xlabel('Подпись оси X')
axes[0, 0].set_ylabel('Подпись оси Y')

axes[0, 1].plot(x, np.cos(x))
axes[0, 1].set_title('cos(x)')
axes[0, 1].grid()

axes[1, 0].plot(x, np.exp(-x),label="№1")
axes[1, 0].plot(x, np.exp(-0.5*x),label="№2")
axes[1, 0].set_title('exp(-x)')
axes[1, 0].legend()

axes[1, 1].plot(x, x**2)
axes[1, 1].set_title('x^2')

fig.suptitle("Набор всяких графиков")
plt.tight_layout()         # автоматически настраивает положение элементов (меток осей, заголовков),
                           # чтобы они не перекрывались и не вылезали за пределы окна фигуры.
plt.show()
```

```python
# Пример создания двумерной сетки и отображения 3D-графика
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm

x = y = np.linspace(-3, 3, 150)
X,Y=np.meshgrid(x, y)
#print(X, Y)
#print(X.shape)
Z = X**2+Y**2

ax = plt.axes(projection='3d')
ax.plot_surface(X,Y,Z)
#ax.plot_surface(X, Y, Z, rstride=4, cstride=4, cmap=cm.jet)
#ax.contourf3D(X, Y, Z, 10)
plt.show()

# Контурная 2D-проекция + colorbar
#plt.contour(X, Y, Z, 40,cmap=cm.jet)
#plt.colorbar()
#plt.show()
```

```python
#Пример: каркас + линии уровня

fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

ax.plot_wireframe(X, Y, Z, rstride=5, cstride=5, color='black', lw=0.5)
#ax.contour(X, Y, Z, zdir='z', offset=-1, cmap='viridis')
ax.contour(X, Y, Z, zdir='z', offset=-1, levels=20, cmap='viridis')
#ax.set_zlim(-1, 1)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('Z')
plt.show()
```
